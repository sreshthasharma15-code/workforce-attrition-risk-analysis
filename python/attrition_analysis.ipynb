```python
department_attrition=round(df['Attrition'].groupby(df['Department']).mean()*100,2).sort_values(ascending=False).reset_index()
```

```python
department_attrition.columns=['Department','AttritionRate']
```

```python
department_attrition
```

```python
fig=px.bar(department_attrition,x='AttritionRate',y='Department',color='Department',title='Department Attrition Rate (%)')
fig.update_layout(xaxis_title='Attrition Rate (%)',showlegend=False)
fig.show()
```

```python
age_bins=[18,25,35,45,55,65]
age_labels=['18-25','26-35','36-45','46-55','56-65']
```

```python
df['AgeGroup']=pd.cut(df['Age'],bins=age_bins,labels=age_labels,include_lowest=True)
```

```python
df['AgeGroup'].value_counts().sort_index()
```

```python
age_wise_attrition=round(df['Attrition'].groupby(df['AgeGroup']).mean()*100,2).reset_index()
```

```python
age_wise_attrition.columns=['AgeGroup','Attrition Rate']
```

```python
age_wise_attrition
```

```python
fig = px.bar(age_wise_attrition,x='AgeGroup',y='Attrition Rate',color='AgeGroup',title='Attrition Rate by Age Group (%)')
fig.update_layout(yaxis_title='Attrition Rate (%)',showlegend=False)
fig.show()
```

```python
tenure_bins=[-1,2,5,10,float('inf')]
tenure_labels=['0-2 Early tenure','3-5 Developing','6-10 Established','11+ Long tenure']
df['TenureGroup']=pd.cut(
    df['YearsAtCompany'],
    bins=tenure_bins,
    labels=tenure_labels
)
```

```python
attrition_by_tenure=round(df['Attrition'].groupby(df['TenureGroup']).mean()*100,2).reset_index()
```

```python
attrition_by_tenure
```

```python
attrition_by_tenure.columns=['Tenure Group','Attrition Rate (%)']
```

```python
fig=px.figsize=(10,10)
fig=px.bar(attrition_by_tenure,x='Tenure Group',y='Attrition Rate (%)',color='Tenure Group',title='Attrition Rate by Tenure')
fig.update_layout(showlegend=False)
fig.show()
```
